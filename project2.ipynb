```python
import pandas as pd
import plotly.express as px

df = pd.read_csv('trade.csv')
```

```python
df.head()
df['Country Code'].unique()
```

```python
gb = df[df['Country Code'] == 'GBR']
gb
birth_year = gb[gb['Year'] == 2002]
birth_year
```

```python
gb.columns
```

```python
import plotly.express as px

gb_sorted = gb.sort_values(by='Year')

fig = px.line(
    gb_sorted,
    x='Year',
    y='average_value_Merchandise exports (current US$)',
    title='The Rise of UK Exports in the past 60 Years',
    labels={'average_value_Merchandise exports (current US$)': 'Annual Value of Exports (US$)'}
)

fig.update_traces(line=dict(color='#00C853'))  # Stock gain green

fig.update_layout(
    plot_bgcolor='black',
    paper_bgcolor='black',
    font=dict(color='white'),
    title_font=dict(color='white'),
    xaxis=dict(color='white', showgrid=False),  # Remove x-axis gridlines
    yaxis=dict(color='white', showgrid=False)   # Remove y-axis gridlines
)

fig.show()
```

```python
# Filter the data for years 2013 and after
gb_filtered = gb_sorted[gb_sorted['Year'] >= 2013]

# Create the plot
fig = px.line(
    gb_filtered,
    x='Year',
    y='average_value_Merchandise exports (current US$)',
    title='The Fall of a Nation: Britain is producing and exporting less goods',
    labels={'average_value_Merchandise exports (current US$)': 'Annual Value of Exports (US$)'}
)

# Update line color to red
fig.update_traces(line=dict(color='red'))  

# Update layout: black background, white text, and remove gridlines
fig.update_layout(
    plot_bgcolor='black',
    paper_bgcolor='black',
    font=dict(color='white'),
    title_font=dict(color='white'),
    xaxis=dict(color='white', showgrid=False),  # Remove x-axis gridlines
    yaxis=dict(color='white', showgrid=False, rangemode='tozero', range=[350000000000, None])  # Y-axis starting at 250 billion
)

fig.show()
```

```python
# Filter the data for years 2013 and after
gb_filtered = gb_sorted[gb_sorted['Year'] >= 2013]

# Create the plot
fig = px.line(
    gb_filtered,
    x='Year',
    y='average_value_Merchandise exports (current US$)',
    title='The Fall of a Nation: Britain is producing and exporting less goods',
    labels={'average_value_Merchandise exports (current US$)': 'Annual Value of Exports (US$)'}
)

# Update line color to red
fig.update_traces(line=dict(color='red'))  

# Update layout: black background, white text, and remove gridlines
fig.update_layout(
    plot_bgcolor='black',
    paper_bgcolor='black',
    font=dict(color='white'),
    title_font=dict(color='white'),
    xaxis=dict(
        color='white', 
        showgrid=False,  # Remove x-axis gridlines
        tickmode='array',
        tickvals=gb_filtered['Year'].unique(),  # Ensure each year gets its own label
        tickangle=0,  # Angle the ticks to prevent overlap
        ticks="inside",  # Move the ticks inside the axis
        ticklen=10,  # Increase the tick length for better readability
        tickwidth=1  # Make the ticks a bit thicker for better visibility
    ),
    yaxis=dict(
        color='white', 
        showgrid=False, 
        rangemode='normal',  # Make sure the range is respected
        range=[350000000000, gb_filtered['average_value_Merchandise exports (current US$)'].max()]  # Y-axis starts at 350 billion
    ),
    margin=dict(l=50, r=50, t=100, b=100)  # Adjust margins for space
)

fig.show()
```
